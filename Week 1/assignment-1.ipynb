```python
img_gray=cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(img_gray,cmap='Greys')
ax.axis('off')
ax.set_title('Gray image')
plt.show()
```

```python
img_resized=cv2.resize(img,None,fx=0.25,fy=0.25)
fig,axs=plt.subplots(figsize=(8,8))
axs.imshow(img_resized)
axs.axis('off')
axs.set_title('Resized onee')
plt.show()
```

```python
img_resize=cv2.resize(img,(100,200))
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(img_resize)
ax.axis('off')
ax.set_title('Resized one')
plt.show()
```

```python
img_extended=cv2.resize(img,(5000,5000),interpolation=cv2.INTER_CUBIC)
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(img_extended)
ax.axis('off')
ax.set_title('Extended one')
plt.show()
```

```python
img_extended=cv2.resize(img,(5000,5000),interpolation=cv2.INTER_LINEAR)
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(img_extended)
ax.axis('off')
ax.set_title('Extended one')
plt.show()
```

```python
kernel_sharpening = np.array([[-1,-1,-1],
                            [-1,9,-1],
                            [-1,-1,-1]])
sharpened=cv2.filter2D(img,-1,kernel_sharpening)

fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(sharpened)
ax.axis('off')
ax.set_title('Sharpened one')
plt.show()
```

```python
import pandas as pd
import numpy as np
from glob import glob
import cv2
import matplotlib.pylab as plt
```

```python
from google.colab import files
uploaded = files.upload()
```

READING IN IMAGES

```python
cat_files = []
dog_files = []

for filename in uploaded.keys():
    lower_name = filename.lower()
    if 'cat' in lower_name:
        cat_files.append(filename)
    elif 'dog' in lower_name:
        dog_files.append(filename)

print("Cat files:", cat_files)
print("Dog files:", dog_files)


# cat_files = glob('C:/GANFORGE/images/cats/*jpg') + glob('C:/GANFORGE/images/cats/*jpeg')
# dog_files = glob('C:/GANFORGE/images/dogs/*jpg') + glob('C:/GANFORGE/images/dogs/*jpeg')

image_mpl = plt.imread(cat_files[3])
image_cv2 = cv2.imread(cat_files[3])
print(len(cat_files))

print(image_mpl.shape)
print(image_cv2.shape)
# # #gives ht, width, channels. (channels for RGB = 3, grayscal = 1 and so on)

pd.Series(image_mpl.flatten()).plot(kind = 'hist', bins = 50, title = 'dist of pixel values')
plt.show()
```

DISPLAYING IMAGES

```python
fig, ax = plt.subplots(figsize = (10,10))
ax.imshow(image_mpl)
ax.axis('off')
plt.show()
```

MATPLOTLIB VS CV2 NUMPY ARRAYS

```python
#cv2 reads in channels as BGR
#matplotlib reads in channels as RGB

fig, axs = plt.subplots(1, 3, figsize = (10, 5))
axs[0].imshow(image_cv2)
axs[1].imshow(image_mpl)
axs[0].axis('off')
axs[1].axis('off')
axs[0].set_title('CV2 image')
axs[1].set_title('matplotlib image')
plt.show()
```

#converting from BGR to RGB

```python
image_cv2_rgb = cv2.cvtColor(image_cv2, cv2.COLOR_BGR2RGB)
fig, ax = plt.subplots()
ax.imshow(image_cv2_rgb)
ax.axis('off')
plt.show()
```

IMAGE MANIPULATION

```python
image = plt.imread(dog_files[0])
fig, ax = plt.subplots(figsize = (8, 8))
ax.imshow(image)
ax.axis('off')
plt.show()

image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
fig, ax = plt.subplots(figsize = (8, 8))
ax.imshow(image_gray, cmap = 'Greys')
ax.axis('off')
ax.set_title('Grey Image')
plt.show()
```

RESIZING AND SCALING

```python
image_resized = cv2.resize(image, None, fx = 0.25, fy = 0.25)
fig, axs = plt.subplots(figsize = (8, 8))
axs.imshow(image_resized)
axs.axis('off')
plt.show()
#different
image_resized = cv2.resize(image, (100, 200))
fig, axs = plt.subplots(figsize = (8, 8))
axs.imshow(image_resized)
axs.axis('off')
plt.show()
#diif
image_resized = cv2.resize(image, (5000, 5000), interpolation = cv2.INTER_CUBIC)
fig, axs = plt.subplots(figsize = (8, 8))
axs.imshow(image_resized)
axs.axis('off')
plt.show()
```

sharpen image

```python
kernel_sharpening = np.array([
    [-1, -1, -1],
    [-1,  9, -1],
    [-1, -1, -1]
], dtype=np.float32)
sharpened = cv2.filter2D(image, -1, kernel_sharpening)
fig, axs = plt.subplots(figsize = (8, 8))
axs.imshow(image_resized)
axs.axis('off')
ax.set_title('sharpened image')
plt.show()
```

blurring the image

```python
kernel_3x3 = np.ones((3, 3),np.float32) / 30
blurred = cv2.filter2D(image, -1, kernel_3x3)
fig, ax = plt.subplots(figsize = (8, 8))
ax.imshow(blurred)
ax.axis('off')
ax.set_title
plt.show()
```

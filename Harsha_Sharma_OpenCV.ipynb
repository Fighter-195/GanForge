```python
# 1. Convert the OpenCV BGR image to RGB so matplotlib displays colors correctly
#    OpenCV loads images in BGR order by default; matplotlib expects RGB.
img_cv2_rgb = cv2.cvtColor(img_cv2, cv2.COLOR_BGR2RGB)

# 2. Create a figure and a single axes object for plotting
fig, ax = plt.subplots(figsize=(6, 6))

# 3. Display the RGB image on the axes
ax.imshow(img_cv2_rgb)

# 4. Turn off axis ticks and labels for a clean image view
ax.axis('off')

# 5. Render the figure to the screen
plt.show()
```

# Image **Manipulation**

```python
img_gray = cv2.cvtColor(img_cv2, cv2.COLOR_RGB2GRAY)

# 2. Create a matplotlib figure and axis
fig, ax = plt.subplots(figsize=(8, 8))

# 3. Display the grayscale image with a grey colormap
ax.imshow(img_gray, cmap='Greys')

# 4. Turn off the axis ticks/labels for a cleaner look
ax.axis('off')

# 5. Add a title
ax.set_title('Grey Image')

# 6. Render the plot
plt.show()
```

# Resize And Scaling **bold text**

```python
# 1. Resize the image to 25% of its original size (fx=0.25, fy=0.25)
img_resized = cv2.resize(img_cv2, None, fx=0.25, fy=0.25, interpolation=cv2.INTER_AREA)

# 2. Create a matplotlib figure and axis for display
fig, ax = plt.subplots(figsize=(8, 8))

# 3. Show the resized image on the axes
ax.imshow(img_resized)

# 4. Turn off axis ticks and labels for a clean view
ax.axis('off')

# 5. Render the figure
plt.show()
```

# Sharpening **Image**

```python
# 1. Define a sharpening kernel: center weight > sum of negatives sharpens edges
kernel_sharpening = np.array([
    [-1, -1, -1],
    [-1,  9, -1],
    [-1, -1, -1]
])

# 2. Apply the kernel to the image using filter2D
#    src: img (BGR or RGB array), ddepth=-1 preserves input depth
sharpened = cv2.filter2D(img_cv2, ddepth=-1, kernel=kernel_sharpening)

# 3. Create a matplotlib figure/axes to display
fig, ax = plt.subplots(figsize=(8, 8))

# 4. Show the sharpened image
#    If img is BGR, convert to RGB first:
#       sharpened = cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB)
ax.imshow(sharpened)

# 5. Turn off axis ticks/labels
ax.axis('off')

# 6. Add a descriptive title
ax.set_title('Sharpened Image')

# 7. Render the result
plt.show()
```

# Blurring **Image**

```python
# make a simple 3×3 averaging kernel
kernel_3x3 = np.ones((3, 3), np.float32) / 9

# apply blur
blurred = cv2.filter2D(img_cv2, -1, kernel_3x3)

# show it
fig, ax = plt.subplots(figsize=(8, 8))
ax.imshow(blurred)
ax.axis('off')
ax.set_title('Blurred Image')
plt.show()
```
